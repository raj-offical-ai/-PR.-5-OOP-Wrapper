```python
print("=== Python OOP Project:- Employee Management System ===")

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_details(self):
        print("Person Name:", self.name)
        print("Person Age:", self.age)

class Employee(Person):
    def __init__(self, name, age, employee_id, salary):
        super().__init__(name, age)
        self.employee_id = employee_id
        self.salary = salary

    def show_details(self):
        print("Employee Name:", self.name)
        print("Employee Age:", self.age)
        print("Employee ID:", self.employee_id)
        print("Employee Salary:", self.salary)

class Manager(Employee):
    def __init__(self, name, age, employee_id, salary, department):
        super().__init__(name, age, employee_id, salary)
        self.department = department

    def show_details(self):
        print("Manager Name:", self.name)
        print("Manager Age:", self.age)
        print("Manager ID:", self.employee_id)
        print("Manager Salary:", self.salary)
        print("Manager Department:", self.department)
        print()
person = None
employee = None
manager = None


while True:
    print()
    print("--- choose another operation ---")
    print()
    print("choose an opration:-")
    print("1. Create a Person")
    print("2. Create a Employee")
    print("3. Create a Manager")
    print("4. Show Details")
    print("5. Exit")
    print()

    choice = input("Enter your choice:- ")

    if choice == "1":
        print()
        print("--- CREATE PERSON ---")

        name = input("Enter Name: ")
        age = input("Enter Age: ")

        person = Person(name, age)

        print(f"Person created with name: {name} and age: {age}.")
        print()

    elif choice == "2":
        print()
        print("--- CREATE EMPLOYEE ---")

        employee_name = input("Enter Name: ")
        employee_age = input("Enter Age: ")
        employee_id = input("Enter Employee ID: ")
        employee_salary = input("Enter Salary: ")

        employee = Employee(
            employee_name,
            employee_age,
            employee_id,
            employee_salary
        )

        print(
            f"Employee created with name: {employee_name}, "
            f"age: {employee_age}, "
            f"id: {employee_id}, "
            f"salary: {employee_salary}"
        )
        print()

    elif choice == "3":
        print()
        print("--- CREATE MANAGER ---")

        manager_name = input("Enter Name: ")
        manager_age = input("Enter Age: ")
        manager_id = input("Enter Employee ID: ")
        manager_salary = input("Enter Salary: ")
        manager_dept = input("Enter Department: ")

        # Manager object
        manager = Manager(
            manager_name,
            manager_age,
            manager_id,
            manager_salary,
            manager_dept
        )

        print(
            f"Manager created with name: {manager_name}, "
            f"age: {manager_age}, "
            f"id: {manager_id}, "
            f"salary: {manager_salary}, "
            f"department: {manager_dept}"
        )
        print()

    elif choice == "4":
        print()
        print("--- SHOW DETAILS ---")

        if person is not None:
            print("Person Name:", person.name)
            print("Person Age:", person.age)
        else:
            print("Person: Not created")

        print()

        if manager is not None:
            manager.show_details()
        else:
            print("Manager: Not created")

        print()

        if employee is not None:
            employee.show_details()
        else:
            print("Employee: Not created")

        print()

    elif choice == "5":
        print()
        print("Existing the system. All resources have been freed.")
        print()
        print("Goodbye...!")
        break

    else:
        print()
        print("Invalid choice! Please enter 1 to 5.")
        print()
```
